```python
from pathlib import Path
import sys

def find_review_layout(start=Path.cwd()):
    for candidate in (start, *start.parents):
        if (candidate / 'scripts' / 'manual_bad_channel_notebook.py').is_file():
            return candidate, candidate / 'scripts', candidate / 'rfi/manual-bad-channels/chime-frb/zach.json'
        if (candidate / 'manual_bad_channel_notebook.py').is_file():
            return candidate, candidate, candidate / 'zach.json'
    raise RuntimeError('The notebook helper files are missing from this project.')

review_root, helper_dir, map_path = find_review_layout()
sys.path.insert(0, str(helper_dir))
from manual_bad_channel_notebook import ManualBadChannelNotebook
review_root
```
